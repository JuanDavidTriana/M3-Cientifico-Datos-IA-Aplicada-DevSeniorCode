# Notebook 05 - GPT y Generacion de Texto

## Objetivos
- Generar texto con GPT-2.
- Comparar temperatura, top_p y top_k.
- Aplicar prompts de resumen y reescritura.

## Introduccion
GPT es decoder-only y autoregresivo: predice el siguiente token. Los parametros de muestreo controlan creatividad vs coherencia.

```python
from transformers import pipeline, set_seed
from IPython.display import display
import pandas as pd

set_seed(42)
generator = pipeline('text-generation', model='gpt2')
print('GPT-2 listo para generar texto')
```

## 1) Generacion basica

```python
prompt = 'In machine learning, transformers'
salida = generator(prompt, max_new_tokens=40, do_sample=False)
print(salida[0]['generated_text'])
```

## 2) Comparacion de temperatura

```python
temperaturas = [0.3, 0.7, 1.2]
filas = []
for temp in temperaturas:
    out = generator(
        'The future of NLP is',
        max_new_tokens=25,
        do_sample=True,
        temperature=temp,
    )
    filas.append({'temperatura': temp, 'texto': out[0]['generated_text']})

display(pd.DataFrame(filas))
```

## 3) Top-p (nucleus) vs Top-k

```python
base = 'Artificial intelligence can help companies'
configs = [
    {'nombre': 'top_k=10', 'top_k': 10, 'top_p': 1.0},
    {'nombre': 'top_p=0.9', 'top_k': 0, 'top_p': 0.9},
    {'nombre': 'top_p=0.5', 'top_k': 0, 'top_p': 0.5},
]
comparacion = []
for cfg in configs:
    out = generator(
        base,
        max_new_tokens=30,
        do_sample=True,
        temperature=0.8,
        top_k=cfg['top_k'],
        top_p=cfg['top_p'],
    )
    comparacion.append({'config': cfg['nombre'], 'texto': out[0]['generated_text']})

display(pd.DataFrame(comparacion))
```

## 4) Prompt de resumen

```python
documento = (
    'Transformers replaced recurrent layers with self-attention, '
    'enabling parallel training and better long-range dependencies.'
)
prompt_resumen = f'Summarize in one sentence:\n{documento}\nSummary:'
resumen = generator(prompt_resumen, max_new_tokens=35, temperature=0.5)[0]['generated_text']
print(resumen)
```

## 5) Prompt de reescritura

```python
texto_formal = 'The quarterly results exceeded expectations due to strong demand.'
prompt_rewrite = (
    f'Rewrite for a general audience:\n{texto_formal}\nRewritten:'
)
reescrito = generator(prompt_rewrite, max_new_tokens=40, temperature=0.7)[0]['generated_text']
print(reescrito)
```

## 6) Ejemplos de prompt engineering

```python
prompts = [
    'List three benefits of transformers in NLP:',
    'Explain self-attention to a beginner:',
    'Complete: Attention is all you need because',
]
for p in prompts:
    txt = generator(p, max_new_tokens=30, temperature=0.6)[0]['generated_text']
    print('-' * 60)
    print(txt)
```

## Resultados
Generamos texto con distintos parametros y observamos trade-offs entre diversidad, fluidez y repeticion.

## Conclusiones
GPT-2 es pequeno pero ilustra principios de LLMs. Temperatura alta aumenta creatividad; top_p restringe el nucleo probabilistico de tokens candidatos.
